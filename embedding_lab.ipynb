```python
from sentence_transformers import SentenceTransformer

# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Category names corresponding to each sublist
category_names = [
    "interests",
    "goals",
    "learning_style",
    "social_preferences",
    "work_style",
    "discussion_topics"
]

encoded_users = {}

for user, categories in user_profile_modified.items():

    encoded_profile = {}

    for category_name, texts in zip(category_names, categories):

        # Combine all phrases in the category into one string
        combined_text = " ".join(texts)

        # Generate ONE embedding for the whole category
        embedding = model.encode(combined_text)

        # Store embedding
        encoded_profile[category_name] = embedding

    encoded_users[user] = encoded_profile


# Example access
print(encoded_users["Arjun"]["interests"])
```

```python
import chromadb

# =========================
# CONNECT TO CHROMA
# =========================

client = chromadb.PersistentClient(path="./chroma_db")

collection = client.get_or_create_collection(
    name="user_profiles"
)

# =========================
# STORE USERS
# =========================

for user, embeddings in encoded_users.items():

    # embeddings is already:
    # {
    #   "interests": vector,
    #   "goals": vector,
    #   ...
    # }

    # -----------------------------------
    # COMBINE CATEGORY VECTORS
    # -----------------------------------

    combined_embedding = []

    for vector in embeddings.values():

        combined_embedding.extend(vector.tolist())

    # -----------------------------------
    # STORE IN CHROMA
    # -----------------------------------

    collection.add(
        ids=[user],

        embeddings=[combined_embedding],

        metadatas=[{
            "name": user
        }]
    )

print("Stored successfully.")
```
